# FLINC Audit Environment Guide

This JupyterLab environment is used to run and capture the IMERG DASK experiment with Sciunit Audit. Use the three notebooks in numerical order.

## 1. Run and Capture the Experiment

1. Open `1_IMERGDASK_FLINC_NASA_ESTIM_DEMO_PARALLEL.ipynb`.
2. Select the **Sciunit Audit (Python 3)** kernel.
3. Run every cell from top to bottom.
4. Confirm that the scientific workflow finishes and the final cleanup cell reports `Dask workers stopped.`
5. Shut down the experiment notebook kernel. This finalizes the captured execution as `e1`. Keep the JupyterLab environment running.

## 2. Verify the Audit Capture

1. Open `2_AFTER_CAPTURE_AUDIT_VERIFY.ipynb`.
2. Select the regular **Python 3 (ipykernel)**, not a Sciunit kernel.
3. Run every cell.
4. Confirm that `e1`, four captured NetCDF inputs, generated arrays, and four tracked PNG maps are reported.

## 3. Finish the Audit Capture

Open `2_AFTER_CAPTURE_AUDIT_VERIFY.ipynb` and run every cell. It checks the finalized `e1` capture, displays the generated maps from the Jupyter workspace, and creates the Sciunit share link in its final cell. Copy that link for the Repeat environment. When verification is complete, use **Stop Environment** in the portal.
